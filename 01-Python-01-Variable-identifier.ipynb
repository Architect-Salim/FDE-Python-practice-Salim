# 01 Day - Python - Variable & identifier 

### Identifier : A name used to identify a program element such as variables, Functions, Class, Etc..

### Variables : A name that refer to an object in memory during execution.

## Rules to define a Variable or Identifier :

### Use Letter and Digits and Underscore.
### Never start Identifier with Digit.
### Do no use Python Keyword as Identifier ( Ex. : if, for , etc..)
### Do not use Space or Hyphens in Identifier.
### Better to use Snake_Case (Ex: server_name , SERVER_NAME)
### Avaoid using Python Built-in Functions name as Variable names)

```python
# How to test which WORD is Keyword or Built-Ins:

import keyword
a=keyword
print(a.iskeyword("if"))
print(a.iskeyword("len")) # Len is not a Keyword
```

```python
# Test , how to identifiy the word is Built-ins or not:?

import builtins
a=builtins
print("print" in dir(a)) # it will print as "print" is Built-ins.
print("if" in dir(a)) # it will NOT print as "if" NOT a Built-ins.
```

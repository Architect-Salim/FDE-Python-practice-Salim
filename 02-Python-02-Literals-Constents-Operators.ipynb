### Comparison Operator :  == , != , > , < , >= , <= 


```python
# == Operator : It is used to check the equality of two values. It will return True if both values are equal otherwise it will return False.
a=5
b=10
print(a==b) # False This means 5 is not equal to 10 so it will return False.
print(a==5) # True This means 5 is equal to 5 so it will return True.
```

```python
# != Operator : It is used to check the inequality of two values. 
# It will return True if both values are not equal otherwise it will return False.
a=5
b=10
print(a!=b) # True This means 5 is not equal to 10 so it will return True.
```

```python
# >= Operator : It is used to check the greater than or equal to of two values. 
# It will return True if the first value is greater than or equal to the second value otherwise it will return False.
a=10
b=10
print(a>=b) # True This means 10 is equal to 10 so it will return True.
```

```python
#Example :
x=17
if x>=18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

### True + True = True

```python
#Example 

a=10
b=20
print(a>5 and b >15) # False This means 10 is not greater than or equal to 20 so it will return False.
```

# "or" Operator:
## ONLY ONE condition must be True, then also it will come as True

### True + True = True
### True + False = True
### False + True = True
### False + False = False

```python
#Example 

a=10
b=20
print(a>5 or b >15) # As 10 is greater than 5, the condition is True.
```

```python
#Example 

a=3
b=20
print(a>5 or b >15) # As 3 is not greater than 5, the condition is False. But as 20 is greater than 15, 
#the condition is True. So the final result will be True.
```

```python
#Example 

a=3
b=2
print(a>5 or b >15) # As 3 is not greater than 5, the condition is False. But as 2 is not greater than 15, the condition is False. So the final result will be False.
```
